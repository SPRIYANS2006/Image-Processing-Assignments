```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
filename = list(uploaded.keys())[0]
img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)
#  (filename,0) would also work
```

```python
border_width = int(input("Enter Border Width in pixels: "))

border_color = int(input("Enter Border Color(0-255): "))

height, width = img.shape

new_img = np.full((height+ 2* border_width,width + 2*border_width), border_color, dtype = np.uint8)

new_img[border_width:border_width+height,
        border_width: border_width +width] = img

cv2_imshow(img)
cv2_imshow(new_img)

# cv2.waitKey(0) and cv2.destroyAllWindows() are not needed with cv2_imshow
```
