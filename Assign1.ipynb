```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
filename = list(uploaded.keys())[0]
img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)
```

```python
threshold = np.mean(img)

binary_img = np.where(img > threshold, 255, 0).astype(np.uint8)

# Display images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(binary_img, cmap='gray')
plt.title(f"Binary Image\nMean Threshold = {threshold:.2f}")
plt.axis("off")

plt.show()
```

```python
threshold = int(input("Enter threshold value (0-255): "))

# Convert grayscale image to binary image
binary_img = np.where(img > threshold, 255, 0).astype(np.uint8)

# Display images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(binary_img, cmap='gray')
plt.title(f"Binary Image\nThreshold = {threshold}")
plt.axis("off")

plt.show()
```
