```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
uploaded = files.upload()

filename = list(uploaded.keys())[0]
```

```python
img = cv2.imread(filename)

# Convert BGR to RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
R = img_rgb[:, :, 0]
G = img_rgb[:, :, 1]
B = img_rgb[:, :, 2]

# Convert RGB to Grayscale using mean average
gray_img = ((R.astype(float) + G.astype(float) + B.astype(float)) / 3).astype(np.uint8)

# Display images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original RGB Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gray_img, cmap="gray")
plt.title("Grayscale Image (Mean Average)")
plt.axis("off")

plt.show()
```

```python
R = img_rgb[:, :, 0]
G = img_rgb[:, :, 1]
B = img_rgb[:, :, 2]

# Input weights from user
wR = float(input("Enter weightage for R (0 to 1): "))
wG = float(input("Enter weightage for G (0 to 1): "))
wB = float(input("Enter weightage for B (0 to 1): "))

# Check whether weights are valid
if wR < 0 or wR > 1 or wG < 0 or wG > 1 or wB < 0 or wB > 1:
    print("Invalid weightage! Values must be between 0 and 1.")

elif abs(wR + wG + wB - 1) > 0.0001:
    print("Invalid weightage! Sum of R, G, and B must be equal to 1.")

else:
    # Convert RGB to Grayscale using weighted average
    gray_img = (
        wR * R.astype(float) +
        wG * G.astype(float) +
        wB * B.astype(float)
    ).astype(np.uint8)

    # Display images
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title("Original RGB Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(gray_img, cmap="gray")
    plt.title("Grayscale Image (Weighted)")
    plt.axis("off")

    plt.show()
```
