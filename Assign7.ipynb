Write a program to sharpen the input image using:
1. Laplacian filter
2. Gradient operators of Sobel and Canny edge detector
3. Unsharp masking. To obtain blurred image, use the Gaussian filter of size 5x5
and sigma= 3

```python

#Laplacian Filter
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("input.jpg", 0)

kernel = np.array([[0,-1,0],
                   [-1,4,-1],
                   [0,-1,0]])

padded = np.pad(img.astype(float), 1, mode='constant')
result = np.zeros(img.shape, dtype=np.uint8)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        total = 0

        for x in range(3):
            for y in range(3):
                total += padded[i+x,j+y] * kernel[x,y]

        value = img[i,j] + total
        result[i,j] = np.clip(value, 0, 255)

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")


plt.subplot(1,2,2)
plt.imshow(result, cmap='gray')
plt.title("Laplacian Sharpening")
plt.axis("off")


plt.show()
```

```python
#Sobel operator

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("input.jpg", 0)

gx = np.array([[-1,0,1],
               [-2,0,2],
               [-1,0,1]])

gy = np.array([[-1,-2,-1],
               [0,0,0],
               [1,2,1]])

padded = np.pad(img.astype(float), 1, mode='constant')
result = np.zeros(img.shape, dtype=np.uint8)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        sx = 0
        sy = 0

        for x in range(3):
            for y in range(3):
                sx += padded[i+x,j+y] * gx[x,y]
                sy += padded[i+x,j+y] * gy[x,y]

        magnitude = np.sqrt(sx**2 + sy**2)
        value = img[i,j] + magnitude
        result[i,j] = np.clip(value, 0, 255)

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(result, cmap='gray')
plt.title("Sobel Sharpening")
plt.axis("off")
plt.show()
```

```python
#Canny edge detector

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("input.jpg", 0)

edges = cv2.Canny(img, 100, 200)

result = np.zeros(img.shape, dtype=np.uint8)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        value = int(img[i,j]) + int(edges[i,j])
        result[i,j] = np.clip(value, 0, 255)

plt.subplot(1,3,1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(edges, cmap='gray')
plt.title("Canny Edges")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(result, cmap='gray')
plt.title("Canny Sharpening")
plt.axis("off")

plt.show()
```

```python
#Unsharp masking

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("input.jpg", 0)

n = 5
sigma = 3
p = n // 2

padded = np.pad(img.astype(float), p, mode='constant')

kernel = np.zeros((n,n), dtype=float)

for x in range(n):
    for y in range(n):
        a = x-p
        b = y-p
        kernel[x,y] = np.exp(-(a*a+b*b)/(2*sigma*sigma))

kernel = kernel / np.sum(kernel)

blurred = np.zeros(img.shape, dtype=float)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        total = 0

        for x in range(n):
            for y in range(n):
                total += padded[i+x,j+y] * kernel[x,y]

        blurred[i,j] = total

mask = img.astype(float) - blurred

sharpened = img.astype(float) + mask

sharpened = np.clip(sharpened, 0, 255).astype(np.uint8)
blurred = np.clip(blurred, 0, 255).astype(np.uint8)

plt.subplot(1,3,1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(blurred, cmap='gray')
plt.title("Gaussian Blurred")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(sharpened, cmap='gray')
plt.title("Unsharp Masking")
plt.axis("off")

plt.show()
```
