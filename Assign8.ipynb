Write a program to generate following noises using equations derived
from PDFs of noise distributions. Compare your output with those
generated using inbuilt function imnoise. Plot the histograms of generated
noise to determine the shape of the distribution.
1. Uniform noise
2. Gaussian noise
3. Erlang noise
4. Exponential noise
5. Rayleigh noise

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("input.jpg", 0)

if img is None:
    print("Image not found!")
    exit()

size = img.shape

a = 0.5
b = 3
mu = 0
sigma = 10

u = np.random.uniform(0, 1, size)
u = np.clip(u, 1e-10, 1)

uniform = np.random.uniform(-20, 20, size)

r1 = np.random.uniform(0, 1, size)
r2 = np.random.uniform(0, 1, size)
gaussian = mu + sigma * np.sqrt(-2*np.log(r1)) * np.cos(2*np.pi*r2)

erlang = np.zeros(size)
for i in range(b):
    u = np.random.uniform(0, 1, size)
    erlang += -np.log(np.clip(u, 1e-10, 1)) / a
erlang = erlang - b/a

u = np.random.uniform(0, 1, size)
exponential = -np.log(np.clip(u, 1e-10, 1)) / a
exponential = exponential - 1/a

u = np.random.uniform(0, 1, size)
rayleigh = np.sqrt(-b * np.log(np.clip(u, 1e-10, 1)))
rayleigh = rayleigh - np.sqrt(np.pi*b)/2

noises = [uniform, gaussian, erlang, exponential, rayleigh]

uniform_ref = np.random.uniform(-20, 20, size)
gaussian_ref = np.random.normal(mu, sigma, size)
erlang_ref = np.random.gamma(b, 1/a, size) - b/a
exponential_ref = np.random.exponential(1/a, size) - 1/a
rayleigh_ref = np.random.rayleigh(np.sqrt(b/2), size) - np.sqrt(np.pi*b)/2

references = [uniform_ref, gaussian_ref, erlang_ref,
              exponential_ref, rayleigh_ref]

names = ["Uniform", "Gaussian", "Erlang", "Exponential", "Rayleigh"]

noisy_images = []

for noise in noises:
    noisy = img.astype(float) + noise
    noisy = np.clip(noisy, 0, 255).astype(np.uint8)
    noisy_images.append(noisy)

plt.figure(figsize=(15, 10))

for i in range(5):
    plt.subplot(2, 3, i+1)
    plt.imshow(noisy_images[i], cmap='gray')
    plt.title(names[i] + " Noise")
    plt.axis("off")

plt.tight_layout()
plt.show()

plt.figure(figsize=(15, 10))

for i in range(5):
    plt.subplot(2, 3, i+1)
    plt.hist(noises[i].flatten(), bins=100, density=True,
             alpha=0.5, label="PDF Equation")
    plt.hist(references[i].flatten(), bins=100, density=True,
             histtype='step', linewidth=1.5, label="NumPy Built-in")
    plt.title(names[i] + " Noise Histogram")
    plt.legend()

plt.tight_layout()
plt.show()
```
